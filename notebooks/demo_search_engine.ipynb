# Part 1: Embeddings-based Search Engine Demo

This notebook demonstrates the **embeddings-based code search engine** implementation for Task Part 1.

## Requirements:
1. ✅ Accept a collection of documents on start
2. ✅ Provide an API to search by text query
3. ✅ Use pretrained embeddings from Hugging Face (e5-base-v2)
4. ✅ Use vector storage library (FAISS)
5. ✅ Demonstrate on test samples

## 1. Setup and Imports

````python
import sys
from pathlib import Path

# Add project root to path
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.engine.faiss_engine import FAISSSearchEngine
from data.load_cosqa import CoSQADataLoader
import json
````

## 2. Initialize Search Engine

**Model**: intfloat/e5-base-v2 from Hugging Face  
**Vector Store**: FAISS (Facebook AI Similarity Search)

````python
# Initialize the search engine with pretrained model
print("Initializing search engine with e5-base-v2 embeddings...")
engine = FAISSSearchEngine(model_name="intfloat/e5-base-v2")

print(f"✓ Model loaded: {engine.model_name}")
print(f"✓ Device: {engine.model.device}")
print(f"✓ Embedding dimension: 768")
````

## 3. Load Document Collection

Loading the CoSQA corpus (20,604 Python code snippets)

````python
# Load the document collection
print("Loading CoSQA corpus...")
loader = CoSQADataLoader(cache_dir=project_root / "data" / "cache")
documents = loader.get_all_corpus()

print(f"✓ Loaded {len(documents):,} documents")
print(f"\nExample document:")
print(f"  ID: {documents[0]['id']}")
print(f"  Text (first 200 chars): {documents[0]['text'][:200]}...")
````

## 4. Ingest Documents (Build Index)

**Requirement 1**: Accept collection of documents on start

````python
# Ingest documents into the search engine
print("Building FAISS index...")
engine.ingest(documents)
print(f"\n✓ Index built successfully!")
print(f"✓ Total documents indexed: {len(engine.documents):,}")
print(f"✓ Index size: {engine.index.ntotal:,} vectors")
````

## 5. Search API Demonstration

**Requirement 2**: Provide API to search by text query

### Example 1: String Manipulation Query

````python
query1 = "How to reverse a string in Python?"
print(f"Query: {query1}\n")
print("="*80)

# Search using the API
results = engine.search(query1, top_k=5)

for i, result in enumerate(results, 1):
    print(f"\nResult #{i}:")
    print(f"  Document ID: {result['id']}")
    print(f"  Similarity Score: {result['score']:.4f}")
    print(f"  Code (first 150 chars):\n    {result['text'][:150]}...")
    print("-" * 80)
````

### Example 2: File I/O Query

````python
query2 = "Read JSON file and parse data"
print(f"Query: {query2}\n")
print("="*80)

results = engine.search(query2, top_k=5)

for i, result in enumerate(results, 1):
    print(f"\nResult #{i}:")
    print(f"  Document ID: {result['id']}")
    print(f"  Similarity Score: {result['score']:.4f}")
    print(f"  Code (first 150 chars):\n    {result['text'][:150]}...")
    print("-" * 80)
````

### Example 3: Data Processing Query

````python
query3 = "Sort a list of dictionaries by specific key"
print(f"Query: {query3}\n")
print("="*80)

results = engine.search(query3, top_k=5)

for i, result in enumerate(results, 1):
    print(f"\nResult #{i}:")
    print(f"  Document ID: {result['id']}")
    print(f"  Similarity Score: {result['score']:.4f}")
    print(f"  Code (first 150 chars):\n    {result['text'][:150]}...")
    print("-" * 80)
````

## 6. Load Real Test Queries from CoSQA

Demonstrate on actual test samples from the dataset

````python
# Load test queries
test_df = loader.load_test()
print(f"Loaded {len(test_df)} test queries from CoSQA\n")

# Show 3 random test queries
import random
sample_queries = test_df.sample(n=3, random_state=42)

for idx, row in sample_queries.iterrows():
    query = row['query_text']
    print(f"\n{'='*80}")
    print(f"Test Query: {query}")
    print(f"{'='*80}")
    
    results = engine.search(query, top_k=3)
    
    for i, result in enumerate(results, 1):
        print(f"\n  Result #{i}:")
        print(f"    Score: {result['score']:.4f}")
        print(f"    Code: {result['text'][:200]}...")
    print()
````

## 7. Search Engine API Summary

### Key Features Demonstrated:

1. ✅ **Document Ingestion**: `engine.ingest(documents)` - Accepts collection of documents
2. ✅ **Search API**: `engine.search(query, top_k)` - Returns ranked results
3. ✅ **Pretrained Embeddings**: Using `intfloat/e5-base-v2` from Hugging Face
4. ✅ **Vector Storage**: FAISS IndexFlatIP for exact similarity search
5. ✅ **Batch Processing**: Efficient encoding of large document collections
6. ✅ **Persistence**: Can save/load indexes with `engine.save_index()` / `engine.load_index()`

### API Methods:

```python
# Initialize
engine = FAISSSearchEngine(model_name="intfloat/e5-base-v2")

# Ingest documents
engine.ingest(documents)  # List[Dict[str, str]] with 'id' and 'text'

# Search
results = engine.search(query, top_k=10)
# Returns: List[Dict] with 'id', 'text', 'score'

# Batch search
results = engine.batch_search(queries, top_k=10)

# Save/load
engine.save_index("path/to/index")
engine.load_index("path/to/index")
```

## 8. Performance Statistics

````python
import time

# Measure search speed
test_queries = [
    "How to read a file line by line?",
    "Convert string to integer in Python",
    "Sort list in descending order",
    "Check if string is palindrome",
    "Find maximum value in list"
]

print("Performance Test: Searching 5 queries...\n")
start_time = time.time()

for query in test_queries:
    results = engine.search(query, top_k=10)

end_time = time.time()
elapsed = end_time - start_time

print(f"✓ Total time: {elapsed:.3f} seconds")
print(f"✓ Average time per query: {elapsed/len(test_queries):.3f} seconds")
print(f"✓ Queries per second: {len(test_queries)/elapsed:.1f}")
print(f"\n✓ Corpus size: {len(documents):,} documents")
print(f"✓ Embedding dimension: 768")
print(f"✓ Index type: FAISS IndexFlatIP (exact search)")
````
